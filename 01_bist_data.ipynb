```python
def find_financial_table(tables, search_text):

    candidates = []

    for i, table in enumerate(tables):

        table_text = (
            table
            .astype(str)
            .apply(lambda col: " ".join(col))
            .str.cat(sep=" ")
        )

        if search_text.lower() in table_text.lower():
            candidates.append((i, table.shape))

    return candidates
```

```python
def find_main_table(tables, search_text):

    candidates = []

    for i, table in enumerate(tables):

        table_text = " ".join(
            table.astype(str).fillna("").values.flatten()
        )

        if search_text.lower() in table_text.lower():
            candidates.append(
                {
                    "index": i,
                    "rows": table.shape[0],
                    "cols": table.shape[1]
                }
            )

    if not candidates:
        return None

    candidates = sorted(
        candidates,
        key=lambda x: x["rows"],
        reverse=True
    )

    return candidates[0]
```

```python
def find_main_table(tables, search_text):

    candidates = []

    for i, table in enumerate(tables):

        table_text = " ".join(
            table.astype(str).fillna("").values.flatten()
        )

        if search_text.lower() in table_text.lower():
            candidates.append((i, table))

    if not candidates:
        raise ValueError(
            f"'{search_text}' içeren tablo bulunamadı."
        )

    # Nested küçük tablolar yerine en fazla satıra sahip ana tablo
    _, main_table = max(
        candidates,
        key=lambda x: x[1].shape[0]
    )

    return main_table
```

```python
import requests
import io
import zipfile
import time
import pandas as pd
import numpy as np

from io import StringIO
from pathlib import Path


# =========================================================
# 1. TICKER -> KAP OID
# =========================================================

def get_oid(ticker):

    url = f"https://www.kap.org.tr/tr/api/member/filter/{ticker}"

    response = requests.get(
        url,
        timeout=60
    )

    response.raise_for_status()

    data = response.json()

    if not data:
        return None

    return data[0]["mkkMemberOid"]


# =========================================================
# 2. FINANCIAL VALUE PARSER
# =========================================================

def parse_financial_value(value):

    if pd.isna(value):
        return np.nan

    # Pandas değeri zaten numeric okuduysa
    if isinstance(
        value,
        (int, float, np.integer, np.floating)
    ):
        return value

    value = str(value).strip()

    if value in ["", "-", "nan", "NaN"]:
        return np.nan

    # Örnek:
    # (1.250.000) -> -1250000

    if value.startswith("(") and value.endswith(")"):

        value = value[1:-1]
        value = value.replace(".", "")

        return -int(value)

    # Örnek:
    # 1.250.000 -> 1250000

    value = value.replace(".", "")

    return int(value)


# =========================================================
# 3. FİNANSAL TABLOYU İÇERİĞİNDEN BUL
# =========================================================

def find_main_table(tables, required_items):

    candidates = []

    for i, table in enumerate(tables):

        table_text = " ".join(
            table
            .astype(str)
            .fillna("")
            .values
            .flatten()
        ).lower()

        # Aradığımız finansal kalemlerin
        # tamamı bu tabloda var mı?

        if all(
            item.lower() in table_text
            for item in required_items
        ):

            candidates.append(
                {
                    "index": i,
                    "rows": table.shape[0],
                    "table": table
                }
            )

    if not candidates:

        raise ValueError(
            f"Gerekli kalemleri içeren tablo bulunamadı: "
            f"{required_items}"
        )

    # Nested küçük HTML tabloları yerine
    # en büyük tabloyu seç

    main_table = max(
        candidates,
        key=lambda x: x["rows"]
    )

    return main_table["table"]


# =========================================================
# 4. FİNANSAL KALEMİ TABLODAN AL
# =========================================================

def get_financial_item(df, item_name):

    mask = (
        df[1]
        .astype(str)
        .str.strip()
        .eq(item_name)
    )

    selected_row = df[mask]

    if selected_row.empty:
        return np.nan

    # Cari dönem
    value = selected_row[3].iloc[0]

    return value


# =========================================================
# 5. SUNUM PARA BİRİMİ / ÖLÇEK
# =========================================================

def get_unit_multiplier(tables):

    info_table = tables[0]

    mask = (
        info_table[0]
        .astype(str)
        .str.strip()
        .eq("Sunum Para Birimi")
    )

    if not mask.any():
        return 1, None

    presentation_currency = (
        info_table.loc[mask, 1]
        .iloc[0]
    )

    presentation_currency = str(
        presentation_currency
    ).strip()

    # Örnekler:
    #
    # TL            -> 1
    # 1 TL          -> 1
    # 1.000 TL      -> 1000
    # 1.000.000 TL  -> 1000000

    numeric_part = (
        presentation_currency
        .upper()
        .replace("TL", "")
        .strip()
    )

    if numeric_part == "":
        multiplier = 1

    else:

        numeric_part = (
            numeric_part
            .replace(".", "")
            .replace(",", "")
        )

        multiplier = int(numeric_part)

    return multiplier, presentation_currency


# =========================================================
# 6. YILLIK RAPORU KAP'TAN AL / CACHE'DEN OKU
# =========================================================

def get_annual_report_tables(
    ticker,
    year,
    cache_dir="kap_cache"
):

    cache_path = Path(cache_dir)

    cache_path.mkdir(
        parents=True,
        exist_ok=True
    )

    cache_file = (
        cache_path /
        f"{ticker}_{year}_annual.xls"
    )


    # -----------------------------------------------------
    # CACHE VARSA
    # -----------------------------------------------------

    if cache_file.exists():

        print(
            f"{ticker} - {year}: CACHE"
        )

        annual_data = cache_file.read_bytes()

        annual_file = cache_file.name


    # -----------------------------------------------------
    # CACHE YOKSA KAP
    # -----------------------------------------------------

    else:

        oid = get_oid(ticker)

        if oid is None:

            raise ValueError(
                f"{ticker} için KAP OID bulunamadı."
            )

        url = (
            "https://www.kap.org.tr/tr/api/home-financial/"
            f"download-file/{oid}/{year}/T"
        )

        response = requests.get(
            url,
            timeout=60
        )

        response.raise_for_status()

        zip_file = io.BytesIO(
            response.content
        )


        # ZIP kontrolü

        if not zipfile.is_zipfile(zip_file):

            raise ValueError(
                f"{ticker} - {year}: "
                "KAP'tan gelen dosya ZIP formatında değil."
            )


        # ZIP'i aç

        with zipfile.ZipFile(zip_file, "r") as z:

            files = z.namelist()

            annual_files = [
                file
                for file in files
                if file.endswith("_4.xls")
            ]

            if not annual_files:

                raise ValueError(
                    f"{ticker} - {year} için "
                    "yıllık finansal rapor bulunamadı."
                )

            if len(annual_files) > 1:

                print(
                    f"UYARI: {ticker} - {year} için "
                    f"{len(annual_files)} yıllık rapor bulundu."
                )

            annual_file = annual_files[0]

            with z.open(annual_file) as file:

                annual_data = file.read()


        # Cache'e kaydet

        cache_file.write_bytes(
            annual_data
        )

        print(
            f"{ticker} - {year}: KAP -> CACHE"
        )


    # -----------------------------------------------------
    # HTML TABLOLARINI OKU
    # -----------------------------------------------------

    tables = pd.read_html(
        StringIO(
            annual_data.decode("utf-8")
        )
    )

    return tables, annual_file


# =========================================================
# 7. TEK ŞİRKET - TEK YIL FİNANSALLARI
# =========================================================

def get_company_financials(ticker, year):

    # -----------------------------------------------------
    # RAPOR
    # -----------------------------------------------------

    tables, annual_file = get_annual_report_tables(
        ticker=ticker,
        year=year
    )


    # -----------------------------------------------------
    # SUNUM ÖLÇEĞİ
    # -----------------------------------------------------

    unit_multiplier, presentation_currency = (
        get_unit_multiplier(tables)
    )


    # -----------------------------------------------------
    # FİNANSAL TABLOLARI DİNAMİK BUL
    # -----------------------------------------------------

    balance_sheet = find_main_table(
        tables,
        [
            "TOPLAM VARLIKLAR",
            "TOPLAM ÖZKAYNAKLAR"
        ]
    )

    income_statement = find_main_table(
        tables,
        [
            "Hasılat",
            "DÖNEM KARI (ZARARI)"
        ]
    )

    cash_flow_statement = find_main_table(
        tables,
        [
            "İŞLETME FAALİYETLERİNDEN NAKİT AKIŞLARI"
        ]
    )


    # -----------------------------------------------------
    # BİLANÇO KALEMLERİ
    # -----------------------------------------------------

    balance_items = {

        "CASH":
            "Nakit ve Nakit Benzerleri",

        "RECEIVABLES":
            "Ticari Alacaklar",

        "INVENTORY":
            "Stoklar",

        "CURRENT_ASSETS":
            "TOPLAM DÖNEN VARLIKLAR",

        "TOTAL_ASSETS":
            "TOPLAM VARLIKLAR",

        "CURRENT_LIABILITIES":
            "TOPLAM KISA VADELİ YÜKÜMLÜLÜKLER",

        "NONCURRENT_LIABILITIES":
            "TOPLAM UZUN VADELİ YÜKÜMLÜLÜKLER",

        "TOTAL_LIABILITIES":
            "TOPLAM YÜKÜMLÜLÜKLER",

        "EQUITY":
            "TOPLAM ÖZKAYNAKLAR"
    }


    # -----------------------------------------------------
    # GELİR TABLOSU
    # -----------------------------------------------------

    income_items = {

        "REVENUE":
            "Hasılat",

        "GROSS_PROFIT":
            "TİCARİ FAALİYETLERDEN BRÜT KAR (ZARAR)",

        "OPERATING_PROFIT":
            "ESAS FAALİYET KARI (ZARARI)",

        "NET_INCOME":
            "DÖNEM KARI (ZARARI)"
    }


    # -----------------------------------------------------
    # NAKİT AKIŞ
    # -----------------------------------------------------

    cash_flow_items = {

        "OPERATING_CASH_FLOW":
            "İŞLETME FAALİYETLERİNDEN NAKİT AKIŞLARI"
    }


    # -----------------------------------------------------
    # SONUÇ
    # -----------------------------------------------------

    financial_data = {

        "TICKER": ticker,

        "YEAR": year,

        "KAP_FILE": annual_file,

        "PRESENTATION_CURRENCY":
            presentation_currency,

        "UNIT_MULTIPLIER":
            unit_multiplier
    }


    # -----------------------------------------------------
    # HELPER
    # -----------------------------------------------------

    def add_financial_item(
        name,
        item_name,
        statement
    ):

        raw_value = get_financial_item(
            statement,
            item_name
        )

        parsed_value = parse_financial_value(
            raw_value
        )

        if pd.isna(parsed_value):

            financial_data[name] = np.nan

        else:

            # Bütün parasal değerleri TL'ye çevir

            financial_data[name] = (
                parsed_value *
                unit_multiplier
            )


    # -----------------------------------------------------
    # BİLANÇO
    # -----------------------------------------------------

    for name, item_name in balance_items.items():

        add_financial_item(
            name,
            item_name,
            balance_sheet
        )


    # -----------------------------------------------------
    # GELİR TABLOSU
    # -----------------------------------------------------

    for name, item_name in income_items.items():

        add_financial_item(
            name,
            item_name,
            income_statement
        )


    # -----------------------------------------------------
    # NAKİT AKIŞ
    # -----------------------------------------------------

    for name, item_name in cash_flow_items.items():

        add_financial_item(
            name,
            item_name,
            cash_flow_statement
        )


    return financial_data


# =========================================================
# 8. TEK ŞİRKET - TARİHSEL VERİ
# =========================================================

def get_company_history(
    ticker,
    start_year,
    end_year,
    request_delay=10,
    max_retries=5,
    cache_dir="kap_cache"
):

    results = []

    for year in range(
        start_year,
        end_year + 1
    ):

        cache_file = (
            Path(cache_dir) /
            f"{ticker}_{year}_annual.xls"
        )

        is_cached = cache_file.exists()


        # -------------------------------------------------
        # RETRY
        # -------------------------------------------------

        for attempt in range(max_retries):

            try:

                financial_data = (
                    get_company_financials(
                        ticker=ticker,
                        year=year
                    )
                )

                results.append(
                    financial_data
                )

                print(
                    f"{ticker} - {year}: OK"
                )


                # Cache'den gelmediyse bekle

                if not is_cached:

                    time.sleep(
                        request_delay
                    )

                break


            # -------------------------------------------------
            # HTTP ERROR
            # -------------------------------------------------

            except requests.exceptions.HTTPError as e:

                if (
                    e.response is not None
                    and
                    e.response.status_code == 429
                ):

                    wait_times = [
                        30,
                        60,
                        120,
                        180,
                        300
                    ]

                    wait_time = wait_times[
                        min(
                            attempt,
                            len(wait_times) - 1
                        )
                    ]

                    print(
                        f"{ticker} - {year}: "
                        f"RATE LIMIT -> "
                        f"{wait_time} sn bekleniyor..."
                    )

                    time.sleep(
                        wait_time
                    )

                else:

                    print(
                        f"{ticker} - {year}: "
                        f"ERROR -> {e}"
                    )

                    break


            # -------------------------------------------------
            # DİĞER ERROR
            # -------------------------------------------------

            except Exception as e:

                print(
                    f"{ticker} - {year}: "
                    f"ERROR -> {e}"
                )

                break


    # =====================================================
    # DATAFRAME
    # =====================================================

    df = pd.DataFrame(
        results
    )

    if not df.empty:

        df = (
            df
            .sort_values("YEAR")
            .reset_index(drop=True)
        )

    return df

def validate_financials(data):

    errors = []

    # Kontrol edilecek temel finansal kolonlar
    required_columns = [
        "CASH",
        "RECEIVABLES",
        "INVENTORY",
        "CURRENT_ASSETS",
        "TOTAL_ASSETS",
        "CURRENT_LIABILITIES",
        "NONCURRENT_LIABILITIES",
        "TOTAL_LIABILITIES",
        "EQUITY",
        "REVENUE",
        "GROSS_PROFIT",
        "OPERATING_PROFIT",
        "NET_INCOME",
        "OPERATING_CASH_FLOW"
    ]

    # 1. Missing kontrolü
    for col in required_columns:

        if col not in data or pd.isna(data[col]):
            errors.append(f"{col}_MISSING")


    # 2. Bilanço eşitliği
    if all(
        not pd.isna(data.get(col))
        for col in [
            "TOTAL_ASSETS",
            "TOTAL_LIABILITIES",
            "EQUITY"
        ]
    ):

        balance_difference = (
            data["TOTAL_ASSETS"]
            - (
                data["TOTAL_LIABILITIES"]
                + data["EQUITY"]
            )
        )

        # Yuvarlama farklarına tolerans
        tolerance = max(
            abs(data["TOTAL_ASSETS"]) * 0.001,
            1
        )

        if abs(balance_difference) > tolerance:

            errors.append(
                "BALANCE_SHEET_MISMATCH"
            )


    # 3. Dönen varlık > toplam varlık olamaz
    if (
        not pd.isna(data.get("CURRENT_ASSETS"))
        and
        not pd.isna(data.get("TOTAL_ASSETS"))
    ):

        if (
            data["CURRENT_ASSETS"]
            > data["TOTAL_ASSETS"]
        ):

            errors.append(
                "CURRENT_ASSETS_GT_TOTAL_ASSETS"
            )


    # 4. Kısa vadeli yükümlülük > toplam yükümlülük olamaz
    if (
        not pd.isna(data.get("CURRENT_LIABILITIES"))
        and
        not pd.isna(data.get("TOTAL_LIABILITIES"))
    ):

        if (
            data["CURRENT_LIABILITIES"]
            > data["TOTAL_LIABILITIES"]
        ):

            errors.append(
                "CURRENT_LIABILITIES_GT_TOTAL_LIABILITIES"
            )


    # 5. Sonuç
    is_valid = len(errors) == 0

    return {
        "TICKER": data.get("TICKER"),
        "YEAR": data.get("YEAR"),
        "IS_VALID": is_valid,
        "ERROR_COUNT": len(errors),
        "ERRORS": errors
    }
```
